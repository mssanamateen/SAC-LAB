```python
#find the gross salary
basic=int(input("enter your basic salary"))
DA=int(input("enter DA"))
HRA=int(input("enter HRA"))
PF=(basic*10)/100
Gross=basic+DA+HRA
NetSal=Gross-PF
print(f"Gross salary is:{Gross}")
print(f"Net salary after PF deduction is {NetSal}")
```

```python
#temperature conversion
temp=float(input("enter the temperature"))
print("1.celsius to fahrenheit")
print("2.fahrenheit to celsius")
ch=int(input("enter ur choice"))
if ch==1:
    fahrenheit=(temp*1.8)+32
    print(f"temperature in fahrenheit is {fahrenheit:.2f}")
elif ch==2:
    celsius=(temp-32)/1.8
    print(f"temperature in celsius is {celsius:.2f}")
else:
    print("invalid choice")
```
