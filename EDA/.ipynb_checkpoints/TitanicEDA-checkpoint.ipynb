### 🧾 Titanic Dataset — Column Description

| Column Name | Description | Data Type |
|--------------|-------------|------------|
| **PassengerId** | Unique ID given to each passenger (used for identification only) | Integer |
| **Survived** | Survival status (0 = Did not survive, 1 = Survived) | Integer / Categorical |
| **Pclass** | Ticket class (1 = 1st, 2 = 2nd, 3 = 3rd) representing socio-economic status | Integer / Ordinal |
| **Name** | Full name of the passenger, includes title (Mr., Mrs., Miss, etc.) | String (Object) |
| **Sex** | Gender of the passenger | String (Categorical) |
| **Age** | Age of the passenger in years (may have missing values) | Float |
| **SibSp** | Number of siblings or spouses aboard | Integer |
| **Parch** | Number of parents or children aboard | Integer |
| **Ticket** | Ticket number (can contain both numbers and letters) | String (Object) |
| **Fare** | Passenger fare (ticket price) | Float |
| **Cabin** | Cabin number (many missing values) | String (Object) |
| **Embarked** | Port of embarkation (C = Cherbourg, Q = Queenstown, S = Southampton) | String (Categorical) |

## Univariate Analysis

````python
import seaborn as sns
import matplotlib.pyplot as plt
````

### 📊 Seaborn `countplot()`

The `sns.countplot()` function in Seaborn is used to create a bar plot that shows the counts of observations in each categorical bin using bars. It's a great way to visualize the frequency distribution of a categorical variable.

---

#### 🧮 Syntax
```python
sns.countplot(x='column_name', data=df)

````python
sns.countplot(x='Survived',data=df)
plt.title('Survival Count')
````

````python
sns.countplot(x='Pclass',data=df)
plt.title("Passenger Class Distribution")
````

### 📈 Seaborn `histplot()`

The `sns.histplot()` function in Seaborn is used to plot histograms, which show the distribution of a single variable. It can also display kernel density estimates (KDE) and work with both continuous and discrete data.

---

#### 🧮 Syntax
```python
sns.histplot(data=df, x='column_name')




kde=True adds a smooth density curve to highlight the distribution shape


````python
sns.histplot(df['Age'].dropna(), kde=True)
plt.title("Age Distribution")
````

## Bivariate Analysis

### 🚻 Survival by Gender

This plot shows the count of survivors and non-survivors grouped by gender using a grouped bar chart.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Survival by Gender")
plt.show()

- x='Sex': Sets the x-axis to the 'Sex' column (Male/Female).
- hue='Survived': Adds color grouping based on survival status (0 = did not survive, 1 = survived).
- data=df: Specifies the DataFrame containing the Titanic dataset.
- plt.title(): Adds a descriptive title to the plot.


````python
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Survival by Gender")
````

- hue='Survived': Differentiates survival status within each class.

```markdown
### 📦 Age vs Survival (Box Plot)

This box plot compares the age distribution of survivors and non-survivors.

```python
sns.boxplot(x='Survived', y='Age', data=df)
plt.title("Age vs Survival")
plt.show()

- x='Survived': Groups data by survival status (0 or 1).
- y='Age': Plots the age distribution for each survival group.
- data=df: Uses the Titanic dataset.
- plt.title(): Adds a title to describe the relationship being visualized.

````python
sns.boxplot(x='Survived', y='Age', data=df)
plt.title("Age vs Survival")
````

## Handle Missing Data

````python
df['Embarked'].value_counts(normalize=True)
````

````python
df['Cabin'].value_counts(normalize=True)
````

````python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
````

````python
df.head()
````

````python
df.isnull().sum()
````

````python
df.info()
````

````python
df['Cabin'] = df['Cabin'].fillna(df['Cabin'].mode()[0])
````

````python
df.isnull().sum()
````

## categorical vs categorical

````python
pd.crosstab(df['Sex'], df['Survived'], normalize='index')
sns.countplot(x='Sex', hue='Survived', data=df)
````

````python
sns.boxplot(x='Survived', y='Age', data=df)
sns.violinplot(x='Survived', y='Fare', data=df)
````
