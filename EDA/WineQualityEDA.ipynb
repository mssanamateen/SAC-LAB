```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("winequality-red.csv",sep=";")
df.head()
```

| Column Name | Description | Data Type | Role |
|--------------|--------------|-----------|------|
| **fixed acidity** | Non-volatile acids (e.g., tartaric acid) that do not evaporate easily. Affects wine’s taste and stability. | float | Feature |
| **volatile acidity** | Amount of acetic acid in the wine (gives vinegar taste if high). | float | Feature |
| **citric acid** | Adds freshness and flavor; affects acidity balance. | float | Feature |
| **residual sugar** | Remaining sugar after fermentation; indicates sweetness. | float | Feature |
| **chlorides** | Salt content in the wine. Too much can spoil the taste. | float | Feature |
| **free sulfur dioxide** | Free SO₂ prevents microbial growth and oxidation. | float | Feature |
| **total sulfur dioxide** | Sum of free and bound SO₂. High levels affect aroma. | float | Feature |
| **density** | Related to alcohol and sugar content; higher density → sweeter wine. | float | Feature |
| **pH** | Measure of acidity (lower = more acidic). Typical range: 2.9–4.0. | float | Feature |
| **sulphates** | Contribute to SO₂ levels; act as preservatives and enhance flavor. | float | Feature |
| **alcohol** | Percentage of alcohol by volume (% ABV). | float | Feature |
| **quality** | Wine quality score (integer scale, usually 3–8). | int | **Target / Output** |

- df.hist(...): Automatically creates histograms for each numeric column in the DataFrame.
- bins=15: Divides the data range into 15 intervals for each histogram.
- figsize=(15, 10): Sets the overall figure size for better readability.
- color='crimson': Fills the bars with a deep red color, matching the dataset theme.
- edgecolor='black': Adds black borders to each bar for visual clarity.
- plt.suptitle(...): Adds a centered title above all subplots.


## Univariate analysis

```python
df.hist(bins=15,figsize=(15,10),color='crimson',edgecolor='black')
plt.suptitle("distribution of featurs in red wine dataset",fontsize=16)
plt.show()
```

## distribution of target variable

```python
sns.countplot(x='quality',data=df,hue='quality',palette='Reds')
plt.title("distribution of wine quality")
plt.show()
```

## descriptive statistics summary

- .T: Transposes the output so that each feature becomes a row and each statistic becomes a column.
- [['mean', 'std', 'min', 'max']]: Selects only the most relevant statistics for quick inspection


```python
df.describe().T[['mean','std','min','max']]
```

## Boxplots to detect outliers

```python
plt.figure(figsize=(15,8))
sns.boxplot(data=df, orient="h", palette="Reds")
plt.title("Boxplot for Outlier Detection in Each Feature")
plt.show()
```
