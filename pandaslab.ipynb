```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("SAC_marks_30students.csv")
print(df)
```

```python
#Line Plot
df['SAC_Theory_Marks'].plot(kind="line")
plt.title("Line Plot")
plt.show()
```

```python
#BAR PLOT
df['SAC_Theory_Marks'].plot(kind="bar")
plt.title("Bar Plot")
plt.show()
```

```python
df.plot(kind='bar', x='Roll_No', y='SAC_Theory_Marks')
```

```python
#histogram
df['SAC_Theory_Marks'].plot(kind="hist")
plt.title("histogram")
plt.show()
```

**The box plot shows five key statistical values:**

- Green line in the middle ≈ Median (82) - the middle value when data is sorted
- Bottom of the box ≈ Q1/25th percentile (65) - 25% of students scored below this
- Top of the box ≈ Q3/75th percentile (89) - 75% of students scored below this
- Bottom whisker ≈ **Minimum (50) - lowest mark**
- Top whisker ≈ **Maximum (95) - highest mark**

```python
#Boxplot

df['SAC_Theory_Marks'].plot(kind="box")
plt.title("Box Plot")
plt.show()
```
